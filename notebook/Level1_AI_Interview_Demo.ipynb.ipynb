```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

MODEL_NAME = "Qwen/Qwen2.5-7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

qwen_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype="auto",
    device_map="auto"
)
```

```python
LEVEL = "Fundamentals"

DOMAIN = "Data Structures and Algorithms"

prompt = f"""
You are an experienced technical interviewer.

Generate exactly 10 interview questions.

Domain:
{DOMAIN}

Difficulty:
{LEVEL}

Rules:

- Ask only beginner-level questions.
- Focus on concepts.
- Focus on definitions.
- Ask simple real-world examples.
- Do NOT ask coding questions.
- Do NOT ask implementation questions.
- Do NOT ask algorithms.

Return ONLY the numbered questions.
"""
```

```python
messages = [
    {"role": "user", "content": prompt}
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    text,
    return_tensors="pt"
)

inputs = {
    k: v.to(next(qwen_model.parameters()).device)
    for k, v in inputs.items()
}
```

```python
outputs = qwen_model.generate(
    **inputs,
    max_new_tokens=500,
    temperature=0.7
)
```

```python
generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]

questions = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
)

print(questions)
```

```python
question_list = []

for line in questions.split("\n"):
    line = line.strip()

    if line == "":
        continue

    if line[0].isdigit():
        q = line.split(".", 1)[1].strip()
        question_list.append(q)

print(question_list)
```

```python
reference_answers = []

for question in question_list:

    prompt = f"""
You are an experienced technical interviewer.

Generate the ideal interview answer.

Question:
{question}

Rules:

- Answer should be technically correct.
- Around 80-120 words.
- Easy English.
- Suitable for freshers.
- Return ONLY the answer.
"""

    messages = [
        {"role":"user","content":prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    )

    inputs = {
        k:v.to(next(qwen_model.parameters()).device)
        for k,v in inputs.items()
    }

    outputs = qwen_model.generate(
        **inputs,
        max_new_tokens=200,
        temperature=0.5
    )

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    answer = tokenizer.decode(
        generated,
        skip_special_tokens=True
    )

    reference_answers.append(answer)
```

```python
for i in range(len(question_list)):

    print("="*60)

    print("Question",i+1)
    print(question_list[i])

    print()

    print("Reference Answer")
    print(reference_answers[i])

    print("="*60)
```

```python
candidate_answers = []

for question in question_list:

    print("="*60)
    print("Question:")
    print(question)
    print("="*60)

    answer = input("Candidate Answer: ")

    candidate_answers.append(answer)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
scores = []

for ref, cand in zip(reference_answers, candidate_answers):

    ref_embedding = embedding_model.encode([ref])

    cand_embedding = embedding_model.encode([cand])

    similarity = cosine_similarity(
        ref_embedding,
        cand_embedding
    )[0][0]

    score = round(similarity * 10, 2)

    scores.append(score)
```

```python
for i in range(len(question_list)):

    print("="*60)

    print("Question",i+1)

    print(question_list[i])

    print()

    print("Reference Answer")

    print(reference_answers[i])

    print()

    print("Candidate Answer")

    print(candidate_answers[i])

    print()

    print("Score :",scores[i],"/10")
```

```python
average_score = round(sum(scores)/len(scores),2)

print("="*60)

print("FINAL LEVEL SCORE")

print(average_score,"/10")
```

```python
if average_score >= 7:

    print("PASS")

    print("Proceed to LEVEL 2")

else:

    print("FAIL")

    print("Repeat LEVEL 1")
```
